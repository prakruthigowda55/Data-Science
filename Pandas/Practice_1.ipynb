Indexing

```python
s=pd.Series([12,30,50,70,90])
print(s)
print(s[0])
print(s[4])

print(s[0:2])  #start(indluded):stop (excluded) : step value(values to jump)

print(s[0:4:2])
```

```python
# iloc--> location based indexing (it gives value in that particular location)
s=pd.Series([12,30,50,70,90])
s.iloc[3]
```

```python
# want multiple values
s.iloc[[1,2,4]]
```

```python
# we can change the index name
s=pd.Series([12,30,50,70,90])
s.name="calories"
print(s)

index=["apple","banana","orange","grapes","mango"]
s.index=index
print(s)
```

```python
s["apple"]
```

```python
# loc --> label based indexing 
print(s.loc['grapes'])
print(s.loc[['grapes','mango']])
```

```python
# in label based indexing your start as well as stop value both are included in the output
print(s['apple':'grapes'])
```

create series using dictonary

```python
fruit_protein ={
    "apples":1.1,     # grams of protein
    "mango":2.2,
    "grapes":0.4,
    "banana":2.5,
    "oranges":0.5
}
s2=pd.Series(fruit_protein,name="protien")
print(s2)
```

```python
print(s2['apples':'grapes'])
```

```python
print(s2.loc[['apples','oranges','mango']])
```

Conditional selection

```python
s2>1
```

```python
s2[s2>1]
```

logical operators: and ,or,not

```python
print([(s2>0.5) & (s2<2)])
print(s2[(s2>0.5) & (s2<=2.5)])
```

```python
print(s2[(s2>0.5) | (s2<=2.5)])
```

```python
print(s2[~(s2>1)])
```

Modifying the series

```python
s2=pd.Series(fruit_protein,name="protien")
print(s2)

s2['mango']=1.0
print(s2)
```
