DataFrames

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Alice"],
    "Age": [25, 30, 35, np.nan, 29, 25],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "HR"],
    "Salary": [50000, 60000, 70000, 62000, np.nan, 50000]
}
data
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
print(df.head(2))  # first 2 data will provide
```

```python
print(df.tail(3))  # last 3 data will provide?
```

loc and iloc

```python
print(df.iloc[1:3,0:3]) # rows,cols
```

```python
print(df.loc[1:2,['Age','Department']])
```

```python
#  access only one colume
print(df['Age'])

# multiple columns
print(df[['Age','Name']])
```

```python
# for just to drop it temporerly
df = pd.DataFrame(data)
print(df)
print(df.drop('Age',axis=1,inplace=False))
```
