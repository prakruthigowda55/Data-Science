DataFrames

```python
import pandas as pd
import numpy as np
```

```python
data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Alice"],
    "Age": [25, 30, 35, np.nan, 29, 25],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "HR"],
    "Salary": [50000, 60000, 70000, 62000, np.nan, 50000]
}
data
```

```python
df = pd.DataFrame(data)
print(df)
```

Data Cleaning--> removing null values

```python
df.isnull().sum()  # checking for null values in data frame and there sum
```

```python
# to get just delete the rows with null value
df.dropna(how="any")  #any row that had any null value
```

```python
df.dropna(how="all") # if all the values in any row are null then we drop
```

```python
df
```

```python
# filling the null values
df["Age"].fillna(df["Age"].mean())  # the null value is replaced with the mean value
```

```python
df["Salary"].fillna(df["Salary"].median())
```

```python
# forward fill--> the nan above value is placed in place of null
df["Age"] = df["Age"].ffill()
```

```python
# bsckward fill--> fill the null vslue with the lower value
df["Salary"] = df["Salary"].bfill()
print(df)
```

```python
# replace a particular name of the row
df["Name"]=df["Name"].replace("Charlie","Rose")
df
```

Duplicate Values

```python
df_dup=df[df.duplicated()]
df_dup
```

```python
df=df.drop_duplicates()
df
```

```python
# invalid values
# lambda--> python

df["promoted salary"] = df["promoted salary"].apply(lambda x:x/10 if x>650000 else x)
df
```
